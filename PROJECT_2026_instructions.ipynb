# Project 2026 · Open-data case study

**Python for Data Analysis** · UESTC–ESIGELEC MSc SEDT · groups of 3 · counts for **70%** of the module grade

## The idea

Public administrations publish thousands of datasets. They are real, useful and messy. Your job is the job of a junior
data analyst: take one raw open dataset, make it reliable, answer a question with figures and statistics, and deliver a
clean, documented, **Machine-Learning-ready** dataset for the next team (in your case: the ML course of semester 2).

## Choose one dataset

| | Dataset | Main difficulty | Starter notebook |
|---|---|---|---|
| **A** | Paris bicycle counters (hourly) + Météo-France daily weather | time series, merging two sources | `PROJECT_A_paris_cycling_weather.ipynb` |
| **B** | DVF: every property sale in Gironde in 2024 | one sale = many rows, outliers | `PROJECT_B_gironde_housing.ipynb` |
| **C** | Météo-France daily climate, Gironde, 1950–2026 | long series, missing values, trends | `PROJECT_C_gironde_climate.ipynb` |

At most **4 groups** per dataset. Two groups on the same dataset must ask **different** main questions.

## What you must do (same steps for every dataset)

1. **Question.** One main question and two sub-questions, written before you start.
2. **Data audit.** What is one row? Shape, types, missing values, duplicates, impossible values, time coverage.
3. **Cleaning.** One function `clean_...(raw)` that starts from the raw file, with every decision commented and checked with `assert`.
4. **Exploration.** At least **five** figures. Under each one, one sentence that states what we learn.
5. **Statistics.** At least one **confidence interval** and one **hypothesis test or OLS regression** (statsmodels), with the effect size in real units.
6. **ML-ready dataset.** Define the observation, the target and the features; split train/test (by time when the data is a time series); compute scaling statistics on the training set only; export a CSV and a data dictionary.
7. **Baseline.** One simple reference and a metric: a naive rule or K-Means (TP4), or the perceptron (TP4, optional). Interpret the result. (Machine Learning models come in semester 2.)
8. **Conclusion.** Answer the question, list the limits of the data and of your method, and what you would do next.

## Deliverables

- **The notebook**, which must run with *Kernel → Restart & Run All* in less than 5 minutes, starting from the file in `data/`.
- **A summary of 2–3 pages** (PDF): question, data, method, three key results with their figures, limits.
- **The ML-ready CSV** and its **data dictionary** (column, type, unit, meaning, how it was built).
- **A defence of 6–8 minutes** followed by 3 minutes of questions. Every member presents one part. Then each
  member makes **one small change to the notebook, live** (for example: change a threshold, add a column to a
  table, change a chart) and runs it.

File names: `GroupNN_Dataset_Name1_Name2_Name3.ipynb` (and `.pdf`, `.csv`). Your teacher gives the exact deadline and
the defence order for your group.

## Grading

| Criterion | Points | What we look at |
|---|---|---|
| Data audit and cleaning | 4 | unit of analysis, duplicates, missing and impossible values found **and** handled, decisions justified |
| Correctness and code quality | 4 | the notebook runs from top to bottom, functions, no data leakage, results are reproducible |
| Analysis and statistics | 4 | relevant figures, at least one confidence interval and one test or regression, effect sizes in real units |
| Interpretation and critical thinking | 4 | answers to the question, limits of the data, what could be wrong, next steps |
| Oral defence (**individual**) | 4 | each member presents one part, answers the questions, and makes a small change to the code live |
| **Total** | **20** | |

The first four criteria are marked for the group; the oral defence is marked **for each student**. The project is an
individual practical exam done in groups: **the members of a group can get different grades**, for example when a member
cannot explain or change the part of the code they present.

## Rules

- You may use the course notebooks, the official documentation and the textbooks.
- You may use an AI assistant to **understand** an error or a function. During the defence, each member makes a
  small change to the code live and explains it: comments in the notebook do not replace understanding. Code that
  the group cannot explain or change gets 0 for the matching criterion.
- Copying between groups is forbidden: same code or same text → 0 for both groups.
- Cite your sources (dataset page, documentation, any external code).
