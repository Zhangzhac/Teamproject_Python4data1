## The data

- `dvf_gironde_2024.csv.gz` — DGFiP, *Demandes de valeurs foncières géolocalisées* (Licence Ouverte), every property
  transaction registered in Gironde (33) in 2024. One row = one **part** of a transaction (a lot, a parcel, a building).
  Main columns: `id_mutation` (transaction id), `date_mutation`, `nature_mutation` (sale, exchange...), `valeur_fonciere`
  (price in €, **repeated on every row of the same transaction**), `type_local` (house, flat, outbuilding, commercial),
  `surface_reelle_bati` (m²), `nombre_pieces_principales`, `surface_terrain`, `nom_commune`, `longitude`, `latitude`.
- Documentation: https://www.data.gouv.fr/fr/datasets/demandes-de-valeurs-foncieres-geolocalisees/

### Known traps (there are others: find them)

- Most transactions have several rows: summing `valeur_fonciere` over rows counts the same price several times.
- A sale can include a house **and** a garage (`Dépendance`), or several flats: decide which transactions are 'simple' and comparable.
- Not every row is a sale (`Echange`, `Adjudication`, `Expropriation`...).
- Surfaces are missing for land and outbuildings; some prices are symbolic (a few euros) or huge.
- Price per m² is very skewed: use medians and log scales.
